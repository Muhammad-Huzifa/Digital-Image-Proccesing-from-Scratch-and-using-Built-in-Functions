# Sampling and interpolation

Compare manual nearest-neighbor resizing with OpenCV nearest, bilinear, and bicubic interpolation.

```python
from pathlib import Path
import sys
import numpy as np
import matplotlib.pyplot as plt

ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents)
            if (p / "src/image_processing").is_dir())
sys.path.insert(0, str(ROOT / "src"))
from image_processing import load_sample, show_images

try:
    import cv2 as cv
except ImportError:
    cv = None
    print("OpenCV is unavailable. NumPy examples still run; install requirements.txt for comparisons.")
```

## Downsample

A smaller image has fewer spatial measurements.

```python
gray = load_sample()
small = gray[::4, ::4]
show_images([gray, small], ["Original", "Downsampled"])
```

## Nearest neighbor from scratch

Each output location selects a source pixel. Read the shared implementation to inspect the index calculation.

```python
from image_processing import nearest_neighbor
manual = nearest_neighbor(small, gray.shape[0], gray.shape[1])
show_images([small, manual], ["Small image", "Manual nearest neighbor"])
if cv is not None:
    builtin = cv.resize(small, (gray.shape[1], gray.shape[0]), interpolation=cv.INTER_NEAREST)
    np.testing.assert_array_equal(manual, builtin)
```

## Bilinear and bicubic interpolation

Interpolation estimates intermediate values; it does not recover detail lost during downsampling.

```python
if cv is not None:
    linear = cv.resize(small, (128, 128), interpolation=cv.INTER_LINEAR)
    cubic = cv.resize(small, (128, 128), interpolation=cv.INTER_CUBIC)
    show_images([manual, linear, cubic], ["Nearest", "Bilinear", "Bicubic"])
```
