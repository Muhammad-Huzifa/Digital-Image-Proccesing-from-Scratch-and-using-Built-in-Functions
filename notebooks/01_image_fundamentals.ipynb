# Image fundamentals

Read images as arrays, inspect pixels, crop regions, and compare spatial sampling with intensity quantization.

```python
from pathlib import Path
import sys
import numpy as np
import matplotlib.pyplot as plt

ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents)
            if (p / "src/image_processing").is_dir())
sys.path.insert(0, str(ROOT / "src"))
from image_processing import load_sample, show_images

try:
    import cv2 as cv
except ImportError:
    cv = None
    print("OpenCV is unavailable. NumPy examples still run; install requirements.txt for comparisons.")
```

## Pixels, shape, and channels

An RGB image has height, width, and three color channels. A grayscale image has one intensity per pixel.

```python
gray = load_sample()
rgb = load_sample("color")
print("Grayscale:", gray.shape, gray.dtype, "RGB:", rgb.shape)
show_images([gray, rgb], ["Grayscale", "RGB"], gray=False)
```

## Crop and replicate

Slicing selects a region. Repeating pixels increases image dimensions without adding new detail.

```python
crop = gray[20:75, 20:75]
replicated = np.repeat(np.repeat(crop, 2, axis=0), 2, axis=1)
show_images([crop, replicated], ["Crop", "Repeated pixels"])
```

## Sampling

Taking every second or fourth pixel reduces spatial resolution.

```python
show_images([gray, gray[::2, ::2], gray[::4, ::4]], ["Original", "Every second pixel", "Every fourth pixel"])
```

## Quantization

Quantization reduces the number of intensity levels while keeping the image dimensions.

```python
images = [gray]
titles = ["8-bit original"]
for bits in [6, 4, 2, 1]:
    step = 256 // (2**bits)
    images.append((gray // step) * step)
    titles.append(f"{bits}-bit")
show_images(images, titles)
```
